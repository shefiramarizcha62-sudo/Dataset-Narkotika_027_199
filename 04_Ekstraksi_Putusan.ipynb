9. Ekstraksi PDF menjadi TXT

```python
hasil_ekstraksi = []

for path_pdf in pdf_files:

    # Mengambil tahun dari nama file
    nama_file = os.path.basename(path_pdf)

    tahun = next(
        (t for t in ['2023', '2024', '2025']
         if f'_{t}_' in nama_file),
        None
    )

    if tahun is None:
        print("Tahun tidak dikenali:", nama_file)
        continue

    nama_txt = os.path.splitext(nama_file)[0] + '.txt'

    folder_txt = os.path.join(TXT_DIR, tahun)
    path_txt = os.path.join(folder_txt, nama_txt)

    try:
        with fitz.open(path_pdf) as dokumen:

            jumlah_halaman = len(dokumen)

            teks = '\n'.join(
                halaman.get_text("text")
                for halaman in dokumen
            )

        with open(path_txt, 'w', encoding='utf-8') as file:
            file.write(teks)

        status = 'Berhasil' if teks.strip() else 'Teks Kosong'

        hasil_ekstraksi.append({
            'Tahun': tahun,
            'Nama PDF': nama_file,
            'Nama TXT': nama_txt,
            'Jumlah Halaman': jumlah_halaman,
            'Jumlah Karakter': len(teks),
            'Status': status
        })

    except Exception as e:

        hasil_ekstraksi.append({
            'Tahun': tahun,
            'Nama PDF': nama_file,
            'Nama TXT': nama_txt,
            'Jumlah Halaman': 0,
            'Jumlah Karakter': 0,
            'Status': f'Error: {e}'
        })

df_hasil = pd.DataFrame(hasil_ekstraksi)

display(df_hasil)
```

## 17.9 Pengujian Ekstraksi Amar Putusan dan Barang Bukti

```python
def ekstrak_amar_dan_barang_bukti(teks):

    # 1. Ekstraksi amar putusan
    pola_amar = r"M\s*E\s*N\s*G\s*A\s*D\s*I\s*L\s*I"

    hasil_amar = list(
        re.finditer(pola_amar, teks, re.IGNORECASE)
    )

    bagian_amar = None

    if hasil_amar:
        posisi_awal = hasil_amar[-1].start()
        bagian_amar = teks[posisi_awal:]

        hasil_disclaimer = re.search(
            r"Disclaimer",
            bagian_amar,
            re.IGNORECASE
        )

        if hasil_disclaimer:
            bagian_amar = bagian_amar[:hasil_disclaimer.start()]

    # 2. Pola barang bukti
    pola_bb = (
        r"(?:"
        r"Menyatakan|Menetapkan|Memerintahkan"
        r")\s+barang(?:-barang)?\s+bukti\s+berupa"
        r"|"
        r"Penuntut\s+Umum\s+mengajukan\s+barang(?:-barang)?\s+bukti"
        r"\s+sebagai\s+berikut"
        r"|"
        r"ditemukan\s+(?:adanya\s+)?barang\s+bukti"
        r"\s+berupa"
    )

    hasil_bb = list(
        re.finditer(
            pola_bb,
            teks,
            re.IGNORECASE
        )
    )

    barang_bukti = None

    if hasil_bb:

        # Ambil pola terakhir agar mengutamakan bagian amar akhir
        hasil = hasil_bb[-1]

        awal_bb = hasil.start()
        bagian_bb = teks[hasil.end():]

        # 3. Membatasi bagian barang bukti
        akhir_bb = re.search(
            r"\n\s*(?:"
            r"Menimbang\s*,|"
            r"Memperhatikan\s*,|"
            r"MENGADILI\s*:?|"
            r"M\s*E\s*N\s*G\s*A\s*D\s*I\s*L\s*I|"
            r"Demikianlah?\s+(?:diputuskan|diucapkan)"
            r")",
            bagian_bb,
            re.IGNORECASE
        )

        if akhir_bb:
            bagian_bb = bagian_bb[:akhir_bb.start()]

        barang_bukti = (
            teks[awal_bb:hasil.end()] + bagian_bb
        ).strip()

    return {
        "amar_putusan": bagian_amar.strip() if bagian_amar else None,
        "barang_bukti": barang_bukti
    }



# 4. Pengujian pada tiga dokumen sampel

for doc in documents:

    if doc["nama_file"] in sampel_putusan:

        hasil = ekstrak_amar_dan_barang_bukti(
            doc["isi_putusan"]
        )

        print("\n" + "=" * 80)
        print("FILE:", doc["nama_file"])
        print("=" * 80)

        print("\n--- BARANG BUKTI ---")
        print(hasil["barang_bukti"])

        print("\n--- AMAR PUTUSAN ---")
        print(hasil["amar_putusan"])
```

## 17.10 — Pemeriksaan Kelengkapan Amar Putusan

```python
for doc in documents:

    if doc["nama_file"] in sampel_putusan:

        teks = doc["isi_putusan"]

        hasil = ekstrak_amar_dan_barang_bukti(teks)

        print("\n" + "=" * 80)
        print("FILE:", doc["nama_file"])
        print("=" * 80)

        amar = hasil["amar_putusan"] or ""

        # Periksa status barang bukti
        pola_status = (
            r"Dirampas\s+untuk\s+dimusnahkan|"
            r"Dirampas\s+untuk\s+negara|"
            r"Dikembalikan\s+kepada|"
            r"Tetap\s+terlampir"
        )

        status = list(
            re.finditer(
                pola_status,
                amar,
                re.IGNORECASE
            )
        )

        print("\nJumlah status barang bukti:", len(status))

        for item in status:
            awal = max(0, item.start() - 250)
            akhir = min(len(amar), item.end() + 250)

            print("\n", amar[awal:akhir])

        # Periksa poin amar terakhir
        print("\n--- 1000 KARAKTER TERAKHIR AMAR ---")
        print(amar[-1000:])
```

## 17.11 Pemeriksaan Struktur Identitas Terdakwa dan Tanggal Putusan

```python
pola_tanggal = (
    r"(?:diucapkan|diucapkan dalam sidang|"
    r"diucapkan dalam persidangan)"
)

for doc in documents:

    if doc["nama_file"] in sampel_putusan:

        teks = doc["isi_putusan"]

        print("\n" + "=" * 80)
        print("FILE:", doc["nama_file"])
        print("=" * 80)

        # 1. Pemeriksaan tanggal pengucapan
        print("\n--- TANGGAL PENGUCAPAN ---")

        matches = list(
            re.finditer(
                pola_tanggal,
                teks,
                re.IGNORECASE
            )
        )

        if matches:
            for match in matches[-3:]:
                awal = max(0, match.start() - 300)
                akhir = min(len(teks), match.end() + 700)
                print(teks[awal:akhir])
        else:
            print("Kata kunci pengucapan tidak ditemukan.")

        # 2. Pemeriksaan identitas terdakwa
        print("\n--- IDENTITAS TERDAKWA ---")

        bagian_awal = teks[:20000]

        pola_nama = (
            r"(?im)^\s*(?:Terdakwa\s+[IVX]+\s*)?"
            r"(?:\d+\.\s*)?"
            r"Nama(?:\s+Lengkap)?\s*:?"
        )

        nama_matches = list(
            re.finditer(
                pola_nama,
                bagian_awal
            )
        )

        print("Jumlah label identitas:", len(nama_matches))

        for i, match in enumerate(nama_matches, 1):
            awal = max(0, match.start() - 100)
            akhir = min(len(bagian_awal), match.end() + 350)

            print(f"\nIdentitas ke-{i}:")
            print(bagian_awal[awal:akhir])
```

## 17.12 Ekstraksi Nama Terdakwa, Jumlah Terdakwa, dan Tanggal Putusan

```python
def ekstrak_nama_terdakwa(teks):

    # Ambil bagian awal dokumen sebelum Disclaimer
    bagian_awal = teks.split("Disclaimer")[0][:20000]

    # Cari pola Nama / Nama Lengkap
    pola = (
        r"(?im)^\s*(?:[IVX]+|\d+)?\s*[\.\)]?\s*"
        r"Nama(?:\s+Lengkap)?\s*:?\s*\n?\s*:?\s*"
        r"([^\n]+)"
    )

    hasil = re.findall(pola, bagian_awal)

    nama_terdakwa = []

    for nama in hasil:
        nama = re.sub(r"\s+", " ", nama).strip(" :;.-")

        if nama and nama not in nama_terdakwa:
            nama_terdakwa.append(nama)

    return nama_terdakwa


# Ekstraksi seluruh dokumen
for doc in documents:

    nama = ekstrak_nama_terdakwa(doc["isi_putusan"])

    print("\n", doc["nama_file"])
    print("Nama terdakwa:", nama)
    print("Jumlah terdakwa:", len(nama))
```

```python
# ==========================================
# 17.17 EKSTRAKSI METADATA TERINTEGRASI
# ==========================================

import os
import re
import pandas as pd

# Daftar bulan untuk ekstraksi tanggal
bulan = (
    r"Januari|Februari|Maret|April|Mei|Juni|"
    r"Juli|Agustus|September|Oktober|November|Desember"
)


# 1. Ekstraksi nomor perkara
def ekstrak_nomor_putusan(teks):

    bagian_awal = teks[:10000]

    pola = (
        r"Nomor\s*:?\s*"
        r"(\d+\s*/\s*Pid\.?\s*Sus\s*/\s*\d{4}\s*/\s*PN\.?\s*Sby\.?)"
    )

    hasil = re.search(pola, bagian_awal, re.IGNORECASE)

    if hasil:
        return re.sub(r"\s+", "", hasil.group(1))

    return None


# 2. Ekstraksi tanggal pengucapan putusan
def ekstrak_tanggal_putusan(teks):

    pola_pengucapan = (
        r"(?:diucapkan|diucapkanlah|diucapkan dalam)"
        r".{0,500}?"
        r"(\d{1,2}\s+(?:" + bulan + r")\s+\d{4})"
    )

    hasil = list(
        re.finditer(
            pola_pengucapan,
            teks,
            re.IGNORECASE | re.DOTALL
        )
    )

    if hasil:
        return hasil[-1].group(1)

    return None


# 3. Ekstraksi pasal
def ekstrak_pasal(teks):

    pola = r"Pasal\s+\d+[A-Za-z]?(?:\s+ayat\s*\(\d+\))?"

    hasil = re.findall(pola, teks, re.IGNORECASE)

    pasal_unik = []
    for item in hasil:
        item = re.sub(r"\s+", " ", item).strip()

        if item.lower() not in [p.lower() for p in pasal_unik]:
            pasal_unik.append(item)

    return pasal_unik


# 4. Ekstraksi seluruh metadata
hasil_final = []

for doc in documents:

    teks = doc["isi_putusan"]
    nama_file = doc["nama_file"]

    # Nomor perkara
    nomor = ekstrak_nomor_putusan(teks)

    # Tanggal pengucapan
    tanggal = ekstrak_tanggal_putusan(teks)

    # Nama terdakwa yang sudah dikoreksi
    nama = doc.get("nama_terdakwa", [])

    # Pasal
    pasal = ekstrak_pasal(teks)

    # Amar dan barang bukti
    hasil_amar = ekstrak_amar_dan_barang_bukti(teks)

    amar = hasil_amar["amar_putusan"]
    barang_bukti = hasil_amar["barang_bukti"]

    # Simpan hasil
    hasil_final.append({

        "case_id": os.path.splitext(nama_file)[0],

        "no_perkara": nomor,

        "tanggal": tanggal,

        "pasal": "; ".join(pasal) if pasal else None,

        "pihak": "; ".join(nama) if nama else None,

        "jumlah_terdakwa": len(nama),

        "ringkasan_fakta": None,

        "argumen_hukum": None,

        "text_full": teks,

        "solusi_text": amar,

        "barang_bukti": barang_bukti,

        "status_nomor": "Terdeteksi" if nomor else "Perlu diperiksa",

        "status_tanggal": "Terdeteksi" if tanggal else "Perlu diperiksa",

        "status_amar": "Terdeteksi" if amar else "Perlu diperiksa",

        "status_barang_bukti": (
            "Terdeteksi" if barang_bukti else "Perlu diperiksa"
        )
    })


# 5. Membuat DataFrame final
df_final = pd.DataFrame(hasil_final)

print("TOTAL DOKUMEN:", len(df_final))

print("\nHASIL DETEKSI METADATA")
for kolom in [
    "status_nomor",
    "status_tanggal",
    "status_amar",
    "status_barang_bukti"
]:
    print(
        kolom, ":",
        (df_final[kolom] == "Terdeteksi").sum(),
        "/", len(df_final)
    )

display(
    df_final[
        [
            "case_id",
            "no_perkara",
            "tanggal",
            "pasal",
            "pihak",
            "jumlah_terdakwa",
            "status_nomor",
            "status_tanggal",
            "status_amar",
            "status_barang_bukti"
        ]
    ]
)
```
