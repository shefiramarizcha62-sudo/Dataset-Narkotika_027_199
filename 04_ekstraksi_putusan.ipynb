```python
import fitz
import os
```

```python
hasil_ekstraksi = []

for tahun in ['2023', '2024', '2025']:

    folder_pdf = os.path.join(PDF_DIR, tahun)
    folder_txt = os.path.join(TXT_DIR, tahun)

    os.makedirs(folder_txt, exist_ok=True)

    file_pdf = sorted([
        f for f in os.listdir(folder_pdf)
        if f.lower().endswith('.pdf')
    ])

    print(f"\nMemproses tahun {tahun}: {len(file_pdf)} PDF")

    for nama_file in file_pdf:

        path_pdf = os.path.join(folder_pdf, nama_file)
        nama_txt = os.path.splitext(nama_file)[0] + '.txt'
        path_txt = os.path.join(folder_txt, nama_txt)

        try:
            with fitz.open(path_pdf) as dokumen:
                jumlah_halaman = len(dokumen)

                daftar_teks = []

                for nomor_halaman, halaman in enumerate(dokumen, start=1):
                    teks_halaman = halaman.get_text("text")

                    daftar_teks.append(
                        f"\n\n========== HALAMAN {nomor_halaman} ==========\n\n"
                        + teks_halaman
                    )

                teks = ''.join(daftar_teks)

            with open(path_txt, 'w', encoding='utf-8') as file:
                file.write(teks)

            status = 'Berhasil' if teks.strip() else 'Teks Kosong'

            hasil_ekstraksi.append({
                'Tahun': tahun,
                'Nama PDF': nama_file,
                'Nama TXT': nama_txt,
                'Jumlah Halaman': jumlah_halaman,
                'Jumlah Karakter': len(teks),
                'Status': status
            })

        except Exception as e:
            hasil_ekstraksi.append({
                'Tahun': tahun,
                'Nama PDF': nama_file,
                'Nama TXT': nama_txt,
                'Jumlah Halaman': 0,
                'Jumlah Karakter': 0,
                'Status': f'Gagal: {str(e)}'
            })

df_hasil = pd.DataFrame(hasil_ekstraksi)
display(df_hasil)
```

```python
import os

tahun_list = ['2023', '2024', '2025']

for tahun in tahun_list:
    folder_pdf = os.path.join(PDF_DIR, tahun)
    folder_txt = os.path.join(TXT_DIR, tahun)

    jumlah_pdf = len([
        f for f in os.listdir(folder_pdf)
        if f.lower().endswith('.pdf')
    ])

    jumlah_txt = len([
        f for f in os.listdir(folder_txt)
        if f.lower().endswith('.txt')
    ])

    print(f'{tahun}: PDF = {jumlah_pdf}, TXT = {jumlah_txt}')
```

periksa seluruh txt

```python
hasil_validasi = []

for tahun in tahun_list:
    folder_txt = os.path.join(TXT_DIR, tahun)

    for nama_file in os.listdir(folder_txt):
        if not nama_file.lower().endswith('.txt'):
            continue

        path_txt = os.path.join(folder_txt, nama_file)

        with open(path_txt, 'r', encoding='utf-8') as f:
            teks = f.read()

        hasil_validasi.append({
            'Tahun': tahun,
            'Nama TXT': nama_file,
            'Jumlah Karakter': len(teks),
            'Jumlah Halaman': teks.count('========== HALAMAN'),
            'Status': 'Berisi teks' if teks.strip() else 'Kosong'
        })

df_validasi = pd.DataFrame(hasil_validasi)

display(df_validasi)

print('\nRingkasan:')
print(df_validasi['Status'].value_counts())
print('\nTotal TXT:', len(df_validasi))
```

validasi pdf dan txt

```python
import random
import os

random.seed(42)

sampel_pdf = []

for tahun in ['2023', '2024', '2025']:
    folder = os.path.join(PDF_DIR, tahun)

    daftar_pdf = sorted([
        f for f in os.listdir(folder)
        if f.lower().endswith('.pdf')
    ])

    jumlah_sampel = min(2 if tahun != '2025' else 1, len(daftar_pdf))

    pilihan = random.sample(daftar_pdf, jumlah_sampel)

    for nama in pilihan:
        sampel_pdf.append({
            'Tahun': tahun,
            'Nama PDF': nama
        })

df_sampel = pd.DataFrame(sampel_pdf)
display(df_sampel)
```

bandingkan pdf dan txt

```python
hasil_perbandingan = []

for data in sampel_pdf:

    tahun = data['Tahun']
    nama_pdf = data['Nama PDF']

    nama_txt = os.path.splitext(nama_pdf)[0] + '.txt'

    path_pdf = os.path.join(PDF_DIR, tahun, nama_pdf)
    path_txt = os.path.join(TXT_DIR, tahun, nama_txt)

    # Membaca teks PDF
    with fitz.open(path_pdf) as dokumen:
        jumlah_halaman_pdf = len(dokumen)

        teks_pdf = '\n'.join(
            halaman.get_text('text')
            for halaman in dokumen
        )

    # Membaca teks TXT
    with open(path_txt, 'r', encoding='utf-8') as file:
        teks_txt = file.read()

    hasil_perbandingan.append({
        'Tahun': tahun,
        'Nama PDF': nama_pdf,
        'Jumlah Karakter PDF': len(teks_pdf),
        'Jumlah Karakter TXT': len(teks_txt),
        'PDF Terbaca': bool(teks_pdf.strip()),
        'TXT Terbaca': bool(teks_txt.strip()),
        'Jumlah Halaman PDF': jumlah_halaman_pdf,
        'Penanda Halaman TXT': teks_txt.count('========== HALAMAN')
    })

df_perbandingan = pd.DataFrame(hasil_perbandingan)

display(df_perbandingan)
```
